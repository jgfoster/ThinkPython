10. Based on the following code, what does `re.search` do?

```python
import re

def test(string):
  return re.search('ea', string.lower()) is not None

def main():
  print(test("Let's go for a swim")) # False
  print(test('Now is the time for each person to vote')) # True
  print(test('Each of us should do what is right')) # True
  print(test('That is an interesting idea')) # True
  print(test('')) # False

main()
```
