## Model 1: C++ Classes

Each class can have a _constructor_ that initializes attributes of a new object. In C++ the constructor is a method with the same name as the class. The constructor is called automatically when you create a new object.

It is common to separate the _interface_ from the _implementation_ by putting the class definition and method prototypes in one file and the code for the methods in another file.

### Atom.h

````python
%%writefile Atom.h
#include <string>
using namespace std;

class Atom {
private:
    string symbol;
    int atomic;
    double mass;
    int isotope;

public:
    // Constructor
    Atom(string sym, int at_num, double at_mass, int iso = 12);

    // Method to calculate the number of neutrons
    int neutrons();

    // Method to convert grams to moles
    double grams_to_moles(double grams);
};
````

<br><br><br><br>

### Atom.cpp

````python
%%writefile Atom.cpp
#include <iostream>
#include <string>
#include "Atom.h"
using namespace std;

// Constructor
Atom::Atom(string sym, int at_num, double at_mass, int iso) {
    symbol = sym;
    atomic = at_num;
    mass = at_mass;
    isotope = iso;
}

// Method to calculate the number of neutrons
int Atom::neutrons() {
    int number = isotope - atomic;
    cout << symbol << " has " << number << " neutrons" << endl;
    return number;
}

// Method to convert grams to moles
double Atom::grams_to_moles(double grams) {
    double moles = grams / mass;
    cout << grams << " g is " << moles << " moles of " << symbol << endl;
    return moles;
}
````

### main.cpp

````python
%%writefile main.cpp
#include "Atom.h"
using namespace std;

int main() {
    Atom oxygen("O", 8, 15.999, 16);  // create an Atom object for oxygen
    Atom carbon("C", 6, 12.001);      // create an Atom object for carbon

    oxygen.neutrons();
    carbon.neutrons();
    oxygen.grams_to_moles(24);
    carbon.grams_to_moles(24);
    
    // oxygen.grams_to_atoms(24);
    // carbon.grams_to_atoms(24);
    return 0;
}
````

````python
%%script bash
g++ Atom.cpp main.cpp -o main && ./main
````
