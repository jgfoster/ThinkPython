### Biology

Classification in biology, also known as taxonomy, involves organizing and categorizing organisms into groups based on shared characteristics. Here are some examples:

* **Kingdom**: The highest level of classification. Examples include Animalia (animals), Plantae (plants), Fungi, Protista, Archaea, and Bacteria.
* **Phylum**: A level below kingdom. For example, within the Animalia kingdom, Chordata (animals with a backbone) and Arthropoda (insects, spiders, crustaceans) are phyla.
* **Class**: A level below phylum. For instance, Mammalia (mammals) is a class within the Chordata phylum.
* **Order**: A level below class. Examples include Carnivora (meat-eating mammals) and Primates (monkeys, apes, and humans).
* **Family**: A level below order. For example, Felidae (cats) and Canidae (dogs) are families within the Carnivora order.
* **Genus**: A level below family. Examples include Panthera (big cats like lions and tigers) and Canis (dogs, wolves).
* **Species**: The most specific level of classification. Examples include Homo sapiens (humans) and Panthera leo (lion).

### Astronomy

In astronomy, classification helps organize and understand the vast array of celestial objects and phenomena. Here are some common classifications:

* **Stars**: Classified by their spectral type (O, B, A, F, G, K, M) based on temperature and color. For example, our Sun is a G-type star.
* **Planets**: Divided into terrestrial planets (rocky, like Earth and Mars) and gas giants (like Jupiter and Saturn). There are also ice giants (like Uranus and Neptune).
* **Galaxies**: Classified by shape into spiral galaxies (like the Milky Way), elliptical galaxies, and irregular galaxies.

### Numbers

Numbers can be classified into various categories based on their properties. Here are some common classifications:
* **Natural Numbers**: Positive integers starting from 1 (1, 2, 3, ...). Sometimes 0 is included, depending on the context.
* **Whole Numbers**: Natural numbers including 0 (0, 1, 2, 3, ...).
* **Integers**: Whole numbers and their negative counterparts (..., -3, -2, -1, 0, 1, 2, 3, ...).
* **Rational Numbers**: Numbers that can be expressed as a fraction of two integers (e.g., 1/2, 3/4, -5/6). This includes integers and fractions.
* **Irrational Numbers**: Numbers that cannot be expressed as a fraction of two integers (e.g., √2, π). Their decimal expansions are non-repeating and non-terminating.
* **Real Numbers**: All rational and irrational numbers. They can be represented on the number line.
* **Complex Numbers**: Numbers that have a real part and an imaginary part (e.g., 3 + 4i, where i is the imaginary unit).

## Model 4: C++ Objects and Classes

Examine the following code as you develop answers to the following questions.

```python
%%writefile main.cpp
#include <iostream>
#include <string>

class Atom {
public:
    std::string symbol;
    int atomic;
    double mass;
    int isotope;

    int neutrons() {
        int number = isotope - atomic;
        std::cout << symbol << " has " << number << " neutrons" << std::endl;
        return number;
    }

    double grams_to_moles(double grams) {
        double moles = grams / mass;
        std::cout << grams << " g is " << moles << " moles of " << symbol << std::endl;
        return moles;
    }
};

int main() {
    Atom oxygen;  // create an Atom object
    oxygen.symbol = "O";
    oxygen.atomic = 8;
    oxygen.mass = 15.999;
    oxygen.isotope = 16;

    Atom carbon;  // create another Atom object
    carbon.symbol = "C";
    carbon.mass = 12.001;

    oxygen.neutrons();
    oxygen.grams_to_moles(24);
    carbon.grams_to_moles(24);

    return 0;
}
```

C++ has a distinction between _primitive types_ (such as `bool`, `char`, `int`, and `double`) and classes (such as `string`). Classes have explicit properties and behavior, while primitive types just have a single value that can be operated on with certain operators (such as `+` for numbers). In a "pure" OOP language such as Smalltalk, the sytax does not make a distinction between types and classes. _Everything_ is an object, even a Boolean or Integer.
