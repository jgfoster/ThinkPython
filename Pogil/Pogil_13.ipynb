## Model 4: Python's `shelve` Module

A simple way to save data is to write it to a text file, one line at a time.

```python
name = 'Adam'
spouse = 'Eve'
residence = 'Eden'
file = open('data.txt', 'w')
file.write(name + '\n')
file.write(spouse + '\n')
file.write(residence + '\n')
file.close()
```

You can then read the data from the file:

```python
file = open('data.txt', 'r')
name = file.readline().strip()
spouse = file.readline().strip()
residence = file.readline().strip()
file.close()
print('Name:', name)
print('Spouse:', spouse)
print('Residence:', residence)
```

The problem with this approach is that it is very rigid: adding fields anywhere except the end is difficult. And if we want to have a variable number of fields, such as names of children, we need to find a flexible way to handle that. What we want instead is a simple way to read and write Python data structures like lists and dictionaries, that themselves can have data structures as elements or values. This is provided by the `shelve` module.
