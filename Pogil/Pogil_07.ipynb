## Model 3: Searching a Sequence

Consider the following code:

```python
def is_odd(number):
  flag = number % 2 == 1
  if flag == True:
    return True # we found an odd number
  else:
    return False # the number was not odd
  
def main():
  for each in range(10):
    if is_odd(each):
      print(each, end = ' ')

main() # should print 1 3 5 7 9
```

## Model 4: Prime Numbers

A prime number is a natural number greater than 1 that is only evenly divisible by 1 and the number itself. For example:
* 2, 3, 5, 7, 11, and 13 are prime numbers.
* 4, 6, 8, 9, and 10 are not prime numbers because they have divisors other than 1 and themselves.

A number, `x`, is evenly divisible by another number, `y`, if the remainder of `x ÷ y` is zero.
