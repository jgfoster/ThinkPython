## Model 3

Compare the following two Python programs:

```python
# get a person's name
name = input("Enter your name: ")

# print the name ten times
for i in range(10):
  print(i, name)

print("Done!")
```

```python
# get a person's name
name = input("Enter your name: ")

# print the name ten times
i = 0
while (i < 10):
  print(i, name)
  i = i + 1

print("Done!")
```

We have seen how a `for-in` statement uses a `range` function to iterate over a block of code a specific number of times (much like the Scratch `repeat (_N_)` control block). An alternate way to iterate over a block of code in Python is with a `while` statement (somewhat analogous to the Scratch `repeat until <_B_>` control block).

```python
def main():
  sum = 0
  ask_another = 'y'
  while ask_another == 'y':
    num = int(input("Enter a number: "))
    sum = sum + num
    ask_another = input("Do you want to enter another number? (y/n): ")
  print("The sum of the numbers you entered is", sum)

# call to main
main()
```
