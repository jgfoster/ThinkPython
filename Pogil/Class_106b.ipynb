## Model 1: Swapping

Follow along as we investigate a swap function.

* Run the program. What are the final values of `x` and `y`?
* Change the function parameters to add a `&` before each name (`int &first, int &second`), save, and run the program. What are the final values of `x` and `y`?

```python
%%writefile main.cpp
#include <iostream>
using namespace std;

void swap(int first, int second) {
  std::cout << "  swap() before: first = " << first << ", second = " << y << std::endl;
  int temp = first;
  first = second;
  second = temp;
  std::cout << "  swap() after : first = " << first << ", second = " << y << std::endl;
}

int main() {
  int x = 6, y = 7;
  std::cout << "Before: x = " << x << ", y = " << y << std::endl;
  swap(x, y);
  std::cout << "After : x = " << x << ", y = " << y << std::endl;
}
```

```python
%%script bash
g++ main.cpp -o main && ./main
```

## Model 2: Factorial

Try the following code here and using a debugger.

* Note that the `factorial` function comes after `main`. How does the compiler know about `factorial` when the function `definition` appears after it is used?
* Note that this code doesn't have `using namespace std;`. What is different about the use of `cin` and `cout`? Why does it still work?
* Can you find the use of a *post-decrement* operator? What would happen if you changed it to a *pre-decrement*?
* Can you rewrite the code to use a `for` loop? Is it easier to understant?

```python
%%writefile main.cpp
#include <cmath>
#include <iostream>

// Precondition: n is an integer greater than or equal to one
// Postcondition: returns the value of n! ( n x (n-1) x (n-2) x ... x 1 )
long factorial(int n);

int main() {
  int n;
  std::cin >> n;

  long val = factorial(n);
  std::cout << val;

  return 0;
}

long factorial(int n) {
  long result = 1;
  while (n > 0) {
    result *= n--;
  }
  return result;
}
```

```python
%%script bash
g++ main.cpp -o main && echo "5" | ./main
```

## Model 3: Function Overloading

```python
%%writefile main.cpp
#include <iostream>
#include <string>
using namespace std;

// ================================================
// Fix this program as follows:
//   1. Overload sortThree() and supporting functions
//       to handle the other function calls in main()
//   2. Do not alter the main() function at all.
//   3. Ensure your output matches the following:
//      Sorted: 1, 2, 3
//      Sorted: 3.5, 7.2, 9.6
//      Sorted: Jon, Laura, William
//
//   Stretch Goal: make a sortFour() function that
//                 sorts four integers

// sort function prototypes
void sortThree(int &x, int &y, int &z);

// supporting swap functions
void swap(int &x, int &y);

// the main program
int main() {

  int a1 = 3, b1 = 1, c1 = 2;
  sortThree(a1, b1, c1);
  std::cout << "Sorted: " << a1 << ", " << b1 << ", " << c1 << std::endl;

  double a2 = 7.2, b2 = 9.6, c2 = 3.5;
  sortThree(a2, b2, c2);
  std::cout << "Sorted: " << a2 << ", " << b2 << ", " << c2 << std::endl;

  std::string a3 = "Laura", b3 = "William", c3 = "Jon";
  sortThree(a3, b3, c3);
  std::cout << "Sorted: " << a3 << ", " << b3 << ", " << c3 << std::endl;
}

// function definitions
void sortThree(int &x, int &y, int &z) {
  if (x > y) {
    swap(x, y);
  }
  if (y > z) {
    swap(y, z);
  }
  if (x > y) {
    swap(x, y);
  }
}

// supporting swap functions
void swap(int &x, int &y) {
  int temp = x;
  x = y;
  y = temp;
}
```

```python
%%script bash
g++ main.cpp -o main && ./main
```
