A _tuple_ is a sequence of values, like a list, but cannot be modified (it is _immutable_). A tuple can be created using a literal syntax with parenthesis (while a literal list is created with square brackets). Like other types, a tuple can be created using the type name as a function.

Predict what will be printed by the following code then run it to see if your prediction is accurate.

```python
t = tuple('abc')
print(type(t))
print(len(t))
```

## Model 2: Tuple Assignment

Predict what will be printed by the following code then run it to see if your prediction is accurate.

```python
x = 24
y = 42

# now swap x and y
temp = x
x = y
y = temp
print('x =', x, '; y =', y)
```

In most programming languages the process of swapping two variables requires three lines and a temporary variable. Predict what will be printed by the following code then run it to see if your prediction is accurate.

```python
x = 24
y = 42

# now swap x and y
x, y = y, x
print('x =', x, '; y =', y)
```

From Model 1 you should recognize that the right-hand side of the assignment statement on line 5 is a tuple. What we are showing above is that the left-hand side of an assignment statement can have multiple variables and will "unpack" the tuple, placing each value in a separate variable.

The `values` method in the dictionary class returns an iterable over the values. But sometimes you want both the key and the value. Predict what will be printed by the following code then run it to see if your prediction is accurate.

```python
d = {'one': 1, 'two': 2}

for item in d.items():
    key, value = item
    print(key, '->', value)
```

The `items` method in dictionary returns an iterable over tuples of keys and values. The above code can be simplified as shown in the next code block.

```python
for key, value in d.items():
    print(key, '->', value)
```

Like most programming languages, Python supports returning a single value from a function or method. But, that value can be a tuple, which can use assignment to unpack the tuple's values into separate variables. Predict what will be printed by the following code then run it to see if your prediction is accurate.

```python
def min_max(numList):
    return min(numList), max(numList)

def main():
    low, high = min_max([2, 4, 1, 3])
    print(low, high)

main()
```

```python
def mean(*args):
    return sum(args) / len(args)

def main():
    print(mean(2, 4, 6, 8))

main()
```
