Then download it with the following code. Note the new file, `pg11.txt` that appears in the sidebar. You may need to refresh the file list to see the change (click the clockwise circle arrow).

```python
from os.path import basename, exists

def download(url):
    filename = basename(url)
    if not exists(filename):
        from urllib.request import urlretrieve

        local, _ = urlretrieve(url, filename)
        print("Downloaded " + str(local))
    return filename

download('https://www.gutenberg.org/cache/epub/11/pg11.txt');
```
