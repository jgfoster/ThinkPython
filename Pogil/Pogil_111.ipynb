## Model 2: A C++ Factorial Function

````python
%%writefile main.cpp
#include <iostream>
using namespace std;

int factorial(int n) {
  std::cout << setw(10 - n * 2) << right << ' ' << "n is " << n << std::endl;
  if (n == 0) {  // line 7
    return 1;    // base case
  } else {
    std::cout << setw(10 - n * 2) << right << ' ' << "need factorial of " << (n - 1) << std::endl;
    int answer = factorial(n - 1);  // line 11
    std::cout << setw(10 - n * 2) << right << ' ' << "factorial of " << (n - 1) << " is " << answer << std::endl;
    return n * answer;
  }
}

int main() {
  int answer = factorial(5);
  cout << endl << "5! = " << answer << endl;
  return 0;
}
````

16. Below is a different recursive implementation of the factorial function seen in Model 2.<br><br>
```C++
int factorial(int n) {
  int result;
  if (n == 0) { // base case
    result = 1;
  } else {      // recursion
    int recurse = factorial(n - 1);
    int result = n * recurse;
  }
  return result;
}
```
