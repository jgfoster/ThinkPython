5. How is the following code different from what you wrote in #4?

```python
def is_even(n):
  """
  Returns 'yes' if n is even, 'no' otherwise
  """
  return 'yes' if n % 2 == 0 else 'no'

def main():
  print(is_even(4))
  print(is_even(5))

main()
```

6. Rewrite the following code to use a conditional expression.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)
    
def main():
    print(factorial(5))
    print(factorial(10))

main()
```
