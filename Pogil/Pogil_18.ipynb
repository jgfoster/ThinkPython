## Model 1: Python's `shelve` Module

A simple way to save data is to write it to a text file, one line at a time.

```python
name = 'Adam'
spouse = 'Eve'
residence = 'Eden'
file = open('data.txt', 'w')
file.write(name + '\n')
file.write(spouse + '\n')
file.write(residence + '\n')
file.close()
```

You can then read the data from the file:

```python
file = open('data.txt', 'r')
name = file.readline().strip()
spouse = file.readline().strip()
residence = file.readline().strip()
file.close()
print('Name:', name)
print('Spouse:', spouse)
print('Residence:', residence)
```

The problem with this approach is that it is very rigid: adding fields anywhere except the end is difficult. And if we want to have a variable number of fields, such as names of children, we need to find a flexible way to handle that. What we want instead is a simple way to read and write Python data structures like lists and dictionaries, that themselves can have data structures as elements or values. This is provided by the `shelve` module.

10. Modify this code to work with `candidates`.

```python
import pandas as pd

# Sample DataFrame with a date column
data = {
    'date': ['2024-01-15', '2024-02-20', '2024-01-25', '2024-03-10', '2024-02-15']
}
df = pd.DataFrame(data)

# Convert the 'date' column to datetime
df['date'] = pd.to_datetime(df['date'])

# Extract the month from the 'date' column
df['month'] = df['date'].dt.month

# Count the frequency of each month
month_frequency = df['month'].value_counts().sort_index()

print(month_frequency)
```

15. How is the following code different from what you wrote in #14?

```python
def is_even(n):
  """
  Returns 'yes' if n is even, 'no' otherwise
  """
  return 'yes' if n % 2 == 0 else 'no'

def main():
  print(is_even(4))
  print(is_even(5))

main()
```

16. Rewrite the following code to use a conditional expression.

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
    
def main():
    print(factorial(5))
    print(factorial(10))

main()
```
