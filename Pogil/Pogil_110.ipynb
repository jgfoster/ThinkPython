## Model 3: Input String Stream

```python
%%writefile main.cpp
#include <iostream>
#include <sstream>
#include <string>
using namespace std;

int main() {
    string data = "abc 42 3.14";
    istringstream inSS(data);
    string label;
    int code;
    double value;
    inSS >> label >> code >> value;
    cout << "Label: " << label << endl;
    cout << "Code:  " << code << endl;
    cout << "Value: " << value << endl;
    return 0;
}
```

```python
%%script bash
g++ main.cpp -o main && ./main
```

## Model 5: File Input

```python
%%writefile data.txt
abc 42 3.14
```

```python
%%writefile main.cpp
#include <iostream>
#include <fstream>
using namespace std;

int main() {
    ifstream inFS;          // Input file stream
    inFS.open("data.txt");
    string label;
    int code;
    double value;
    inFS >> label >> code >> value;
    inFS.close();
    cout << "Label: " << label << endl;
    cout << "Code:  " << code << endl;
    cout << "Value: " << value << endl;
    return 0;
}
```

```python
%%script bash
g++ main.cpp -o main && ./main
```
