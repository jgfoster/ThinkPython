## Model 2: Two C++ Code Snippets

Refer to Model 2 below as your team develops consensus answers to the questions below.

````python
%%writefile main.cpp
#include <iostream>
using namespace std;

void double() {
  // put first 100 powers of two in x
  vector<double> x(100);
  for (int i = 0; i < x.size(); i++) {
    x.at(i) = pow(2, i);
  }
}
void sum(vector<int> y) {
  // sum the elements of y
  int sum = 0;
  for (int i = 0; i < y.size(); i++) {
    sum += y.at(i);
  }
}

int main() {
  double();
  return 0;
}
````

````python
%%script bash
g++ main.cpp -o main && ./main
````
